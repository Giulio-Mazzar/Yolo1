# Export the .pt yolo model into .tflite
## Introduction
In order to use YOLO object detection algorithm in android studio, a tflite model must be adopted. Therefore the conversion from standard
` .pt ` file into a ` .tflite ` must be done.



## Install Ultralytics packages

```python
!pip install ultralytics
```

## Export the selected model in .tflite format

```python
from ultralytics import YOLO

# Load the YOLO model
model = YOLO("my_model.pt")

# Export the model to TFLite
model.export(format="tflite")  # Creates 'yolo11n_float32.tflite'
```
